```python
img = cv2.imread('pessi.jpg')
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
img_gray
```

```python
# 1 - lê a imagem
# 2 - converter para cinza
# 3 - percorre a matrix de imagem e, p/ cada pixel
## identificar a faixa de itensidade
## incrementar a posição correspondente no vetor
def calc_hist(img):
  img_vetor = img.ravel()
  hist_vetor = np.zeros(256)

  for px in img_vetor:
    hist_vetor[px] += 1

  return hist_vetor
```

```python
hist_img = calc_hist(img_gray)

plt.bar(range(256), hist_img)
```
