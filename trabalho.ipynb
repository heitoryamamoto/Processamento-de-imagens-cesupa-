```python

img = cv2.imread('EI9fKKSXkAAx4KR.jpg')
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(img_gray, cmap='gray')
plt.title('Imagem em tons de cinza')
plt.axis('off')
plt.show()
```

```python
def quantizar(img_gray, bits):
    niveis = 2 ** bits
    passo = 256 // niveis
    img_quant = (img_gray // passo) * passo
    return img_quant.astype(np.uint8)
```

```python
img_1bit = quantizar(img_gray, 1)
img_2bit = quantizar(img_gray, 2)
img_4bit = quantizar(img_gray, 4)
img_8bit = quantizar(img_gray, 8)
```

```python
fig, ax = plt.subplots(1, 5, figsize=(18, 4))

ax[0].imshow(img_gray, cmap='gray', vmin=0, vmax=255)
ax[0].set_title('Original (cinza)')
ax[0].axis('off')

ax[1].imshow(img_1bit, cmap='gray', vmin=0, vmax=255)
ax[1].set_title('1 bit - 2 niveis')
ax[1].axis('off')

ax[2].imshow(img_2bit, cmap='gray', vmin=0, vmax=255)
ax[2].set_title('2 bits - 4 niveis')
ax[2].axis('off')

ax[3].imshow(img_4bit, cmap='gray', vmin=0, vmax=255)
ax[3].set_title('4 bits - 16 niveis')
ax[3].axis('off')

ax[4].imshow(img_8bit, cmap='gray', vmin=0, vmax=255)
ax[4].set_title('8 bits - 256 niveis')
ax[4].axis('off')

plt.tight_layout()
plt.savefig('quantizacao.png')
plt.show()
```
